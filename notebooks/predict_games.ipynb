```python
import sys

sys.path.append("..")

import joblib
import polars as pl
```

# Predict Games

model_path: str
    Path to a pickled tuple of type (estimator, train_cols)

predict_path: str
    Path to the parquet file containing the data to be predicted.

league_id: int
    Legaue_id of the matches to be predited.
    
match_day: int
    Match day to be predicted.

```python
model_path = "../models/lgbm.pkl"
predict_path = "/home/fabian/Project-Data/AK-Tippspiel/openligadb/openligadb_result_class_predict.parquet"
league_id = 4741
match_day = 13
```

## Load model and train cols

```python
model, train_cols = joblib.load(model_path)
model.estimator
```

## Load data to be predicted

```python
df_predict = (
    pl.scan_parquet(predict_path)
    .filter((pl.col("league_id") == league_id) & (pl.col("match_day") == match_day))
    .collect()
)
df_predict
```

## Prediction

```python
df_predict = df_predict.with_columns(model.predict(df_predict[train_cols]))
df_predict.select(
    ["team_name_1", "team_name_2", "goals_team_1_pred_lgbm", "goals_team_2_pred_lgbm"]
)
```
