```python
import sys

sys.path.append("..")
from aktipp.scraping import scrape_many_seasons_openligadb
from aktipp.normalize import normalize_many_seasons_openligadb
from aktipp.etl import (
    clean_openligadb,
    create_performance_openligadb,
    create_standings_openligadb,
    FeatureBuilderOpenligadb,
)
```

## Parameters

```python
# leagues = ["bl1", "bl2", "bl3"]
leagues = ["bl1"]
# seasons = range(2000, 2025)
seasons = [2024]
data_path = "/home/fabian/Project-Data/AK-Tippspiel/openligadb/"
```

## Dowload, normnalize and clean openligadb data

```python
scrape_many_seasons_openligadb(leagues, seasons, data_path)
normalize_many_seasons_openligadb(leagues, seasons, data_path)
clean_openligadb(data_path)
```

## Create feature data

```python
create_performance_openligadb(
    match_results_data_path=data_path + "matchResults_clean.parquet",
    performance_data_path=data_path + "overall_performance.parquet",
    performance_class="overall",
)
```

```python
create_performance_openligadb(
    match_results_data_path=data_path + "matchResults_clean.parquet",
    performance_data_path=data_path + "home_away_performance.parquet",
    performance_class="home_away",
)
```

```python
create_standings_openligadb(
    match_results_data_path=data_path + "matchResults_clean.parquet",
    standings_data_path=data_path + "overall_standings.parquet",
    standings_class="overall",
)
```

```python
create_standings_openligadb(
    match_results_data_path=data_path + "matchResults_clean.parquet",
    standings_data_path=data_path + "home_standings.parquet",
    standings_class="home",
)
```

```python
create_standings_openligadb(
    match_results_data_path=data_path + "matchResults_clean.parquet",
    standings_data_path=data_path + "away_standings.parquet",
    standings_class="away",
)
```

## Data for class_result

```python
builder = FeatureBuilderOpenligadb()
df = builder.get_features(
    match_results_data_path="/home/fabian/Project-Data/AK-Tippspiel/openligadb/matchResults_clean.parquet",
    features={
        "overall_standings": "/home/fabian/Project-Data/AK-Tippspiel/openligadb/overall_standings.parquet",
        "overall_performance": "/home/fabian/Project-Data/AK-Tippspiel/openligadb/overall_performance.parquet",
    },
    target="result_class",
).sink_parquet(data_path + "openligadb_result_class_dataset.parquet")
```
